Четвертое задание

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 4, 6, 8, 10])
corr = np.corrcoef(a, b)[0, 1]
print(a)
print(' ')
print(b)
print(' ')
print(corr)
```
